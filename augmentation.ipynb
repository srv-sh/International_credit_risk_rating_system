```python
import os
import imgaug.augmenters as iaa
from PIL import Image
import numpy as np

def augment_images(input_folder, output_folder):
    os.makedirs(output_folder, exist_ok=True)
    
    augmentations = [
        iaa.Affine(rotate=5),  # Rotate 5 degrees
        iaa.Affine(rotate=7),  # Rotate 7 degrees
        iaa.MultiplyBrightness(1.5),  # Increase brightness
        iaa.LinearContrast(1.5),  # Increase contrast
        # iaa.GaussianBlur(sigma=1.0),  # Blur to mimic photocopy effect
        iaa.AdditiveGaussianNoise(scale=15),  # Add noise
        iaa.Grayscale(alpha=1.0)  # Convert to grayscale
    ]
    
    for filename in os.listdir(input_folder):
        if filename.lower().endswith(('png', 'jpg', 'jpeg')):
            image_path = os.path.join(input_folder, filename)
            image = np.array(Image.open(image_path))
            
            for i, aug in enumerate(augmentations):
                augmented_image = aug(image=image)
                augmented_img_pil = Image.fromarray(augmented_image)
                new_filename = f"{os.path.splitext(filename)[0]}_aug{i+1}.jpg"
                augmented_img_pil.save(os.path.join(output_folder, new_filename))
                
            print(f"Augmented {filename} and saved to {output_folder}")

# Example usage
input_folder = "images"  # Folder containing original images
output_folder = "augmented_images"  # Folder to save augmented images
augment_images(input_folder, output_folder)
```
